```python
import numpy as np
import sympy as sp

np.set_printoptions(precision=4, suppress=True)
sp.init_printing()

def clean_array(A, tol=1e-12):
    A = np.array(A, dtype=complex)
    A[np.abs(A.real) < tol] = 1j * A[np.abs(A.real) < tol].imag
    A[np.abs(A.imag) < tol] = A[np.abs(A.imag) < tol].real
    return A

def print_ket(name, v):
    v = clean_array(v)
    print(f"|{name}> =")
    print(v)
    print()

#Menghitung inner product <bra|ket>
def inner_product(bra, ket):
    bra = np.array(bra, dtype=complex).flatten()
    ket = np.array(ket, dtype=complex).flatten()
    return np.vdot(bra, ket)  # np.vdot otomatis konjugasi elemen pertama

#Menghitung probabilitas P = |<bra|ket>|^2
def probability(bra, ket):
    return abs(inner_product(bra, ket))**2

#Menghitung norma (panjang) vektor keadaan
def norm(v):
    return np.sqrt(probability(v, v))

# Keadaan basis HV
ket_H = np.array([[1], [0]], dtype=complex)
ket_V = np.array([[0], [1]], dtype=complex)

# Keadaan +-45
ket_plus45  = (1/np.sqrt(2)) * np.array([[1], [ 1]], dtype=complex)
ket_minus45 = (1/np.sqrt(2)) * np.array([[1], [-1]], dtype=complex)

# Keadaan melingkar
ket_L = (1/np.sqrt(2)) * np.array([[1], [ 1j]], dtype=complex)
ket_R = (1/np.sqrt(2)) * np.array([[1], [-1j]], dtype=complex)

#Keadaan terpolarisasi linier dengan sudut theta dari horizontal
def ket_theta(theta_deg):
    theta = np.deg2rad(theta_deg)
    return np.array([[np.cos(theta)], [np.sin(theta)]], dtype=complex)

#Keadaan polarisasi eliptis
def ket_elliptical(theta_deg, phi_deg):
    theta = np.deg2rad(theta_deg)
    phi   = np.deg2rad(phi_deg)
    return np.array([[np.cos(theta)], [np.exp(1j*phi)*np.sin(theta)]], dtype=complex)
```

# Vektor Keadaan dan Basis

Dalam mekanika kuantum, polarisasi foton tidak lagi dinyatakan sebagai vektor polarisasi klasik $\vec{P}$, melainkan sebagai vektor keadaan (ket) dalam notasi Dirac.

Dua keadaan basis ortonormal untuk polarisasi adalah

$$
|H\rangle\doteq\begin{bmatrix}1\\0\end{bmatrix}_{HV},
\qquad
|V\rangle\doteq\begin{bmatrix}0\\1\end{bmatrix}_{HV}.
$$

Sembarang keadaan polarisasi dapat dinyatakan sebagai superposisi (kombinasi linier) dari kedua basis ini:

$$
|\psi\rangle = c_H|H\rangle + c_V|V\rangle,
$$

dengan $c_H$ dan $c_V$ adalah bilangan kompleks.

```python
print_ket("H", ket_H)
print_ket("V", ket_V)
print_ket("+45", ket_plus45)
print_ket("-45", ket_minus45)
print_ket("L", ket_L)
print_ket("R", ket_R)

print("Norma setiap keadaan basis (harus = 1):")
for name, ket in [("H", ket_H), ("V", ket_V), ("+45", ket_plus45),
                  ("-45", ket_minus45), ("L", ket_L), ("R", ket_R)]:
    print(f"  norm(|{name}>) = {norm(ket):.6f}")
```

# Ortogonalitas dan Inner Product

Untuk setiap ket $|\psi\rangle$, terdapat pasangan bra $\langle\psi|$ yang merupakan konjugat transpos dari vektor kolomnya.

Inner product antara dua keadaan kuantum didefinisikan sebagai

$$
\langle\psi_1|\psi_2\rangle = c \in \mathbb{C}.
$$

Karena $|H\rangle$ dan $|V\rangle$ merupakan basis ortonormal, berlaku

$$
\langle H|V\rangle = \langle V|H\rangle = 0 \quad (\text{ortogonal}),
$$

$$
\langle H|H\rangle = \langle V|V\rangle = 1 \quad (\text{ternormalisasi}).
$$

Dari sifat ini, koefisien basis dapat ditemukan melalui proyeksi:

$$
c_H = \langle H|\psi\rangle, \qquad c_V = \langle V|\psi\rangle.
$$

Selain itu, berlaku pula sifat konjugasi:

$$
\langle\psi_1|\psi_2\rangle^* = \langle\psi_2|\psi_1\rangle.
$$

```python
print("Ortonormalitas basis HV")
print(f"<H|H> = {inner_product(ket_H, ket_H)}")
print(f"<V|V> = {inner_product(ket_V, ket_V)}")
print(f"<H|V> = {inner_product(ket_H, ket_V)}")
print(f"<V|H> = {inner_product(ket_V, ket_H)}")

print()
print("Sifat konjugasi: <psi1|psi2>* = <psi2|psi1>")
ip12 = inner_product(ket_plus45, ket_L)
ip21 = inner_product(ket_L, ket_plus45)
print(f"<+45|L>  = {clean_array(ip12)}")
print(f"<L|+45>  = {clean_array(ip21)}")
print(f"<+45|L>* = {clean_array(ip12.conj())}")
print()
print("Apakah <+45|L>* = <L|+45>?")
print(np.isclose(ip12.conj(), ip21))
```

# Kondisi Normalisasi

Untuk keadaan umum $|\psi\rangle = c_H|H\rangle + c_V|V\rangle$, kondisi normalisasi mensyaratkan

$$
\langle\psi|\psi\rangle = |c_H|^2 + |c_V|^2 = 1.
$$

Ini adalah dasar dari interpretasi probabilistik mekanika kuantum dimana $|c_H|^2$ adalah probabilitas foton terukur berada pada keadaan $|H\rangle$, dan $|c_V|^2$ untuk keadaan $|V\rangle$.

```python
print("Verifikasi kondisi normalisasi <psi|psi> = 1 ")
for name, ket in [("H", ket_H), ("V", ket_V), ("+45", ket_plus45),
                  ("-45", ket_minus45), ("L", ket_L), ("R", ket_R)]:
    val = inner_product(ket, ket)
    print(f"  <{name}|{name}> = {val.real:.6f}")

print()
print("Dekomposisi keadaan dalam basis HV")
# Temukan c_H dan c_V untuk keadaan |+45>
c_H = inner_product(ket_H, ket_plus45)
c_V = inner_product(ket_V, ket_plus45)
print(f"|+45> = c_H|H> + c_V|V>")
print(f"  c_H = <H|+45> = {clean_array(c_H)}")
print(f"  c_V = <V|+45> = {clean_array(c_V)}")
print(f"  |c_H|^2 + |c_V|^2 = {abs(c_H)**2 + abs(c_V)**2:.6f}")

print()
# Rekonstruksi |+45> dari koefisiennya
reconstructed = c_H * ket_H + c_V * ket_V
print("Rekonstruksi |+45> dari koefisiennya:")
print_ket("+45 (rekonstruksi)", reconstructed)
print("Apakah rekonstruksi sama dengan |+45> asli?")
print(np.allclose(reconstructed, ket_plus45))
```

# Keadaan $|{\pm}45\rangle$ sebagai Superposisi $|H\rangle$ dan $|V\rangle$

Dari eksperimen (dan analogi klasik), keadaan $|{+}45\rangle$ dan $|{-}45\rangle$ dapat dinyatakan dalam basis $HV$ sebagai

$$
|{+}45\rangle = \frac{1}{\sqrt{2}}\bigl(|H\rangle + |V\rangle\bigr),
$$

$$
|{-}45\rangle = \frac{1}{\sqrt{2}}\bigl(|H\rangle - |V\rangle\bigr).
$$

Kedua keadaan ini saling ortogonal: $\langle{+}45|{-}45\rangle = 0$.

Sebaliknya, keadaan basis $|H\rangle$ dan $|V\rangle$ juga bisa dinyatakan dalam basis $\{|{+}45\rangle, |{-}45\rangle\}$:

$$
|H\rangle = \frac{1}{\sqrt{2}}\bigl(|{+}45\rangle + |{-}45\rangle\bigr),
\qquad
|V\rangle = \frac{1}{\sqrt{2}}\bigl(|{+}45\rangle - |{-}45\rangle\bigr).
$$

```python
print("Ortogonalitas keadaan +-45 ")
ip = inner_product(ket_plus45, ket_minus45)
print(f"<+45|-45> = {clean_array(ip)}")
print(f"<-45|+45> = {clean_array(inner_product(ket_minus45, ket_plus45))}")

print()
print("Verifikasi: |H> dan |V> dalam basis +-45")
ket_H_from_45 = (1/np.sqrt(2)) * (ket_plus45 + ket_minus45)
ket_V_from_45 = (1/np.sqrt(2)) * (ket_plus45 - ket_minus45)

print_ket("H (dari +-45)", ket_H_from_45)
print_ket("V (dari +-45)", ket_V_from_45)

print("Apakah |H> (dari +-45) sama dengan |H> asli?")
print(np.allclose(ket_H_from_45, ket_H))
print("Apakah |V> (dari +-45) sama dengan |V> asli?")
print(np.allclose(ket_V_from_45, ket_V))
```

# Representasi Vektor Baris dan Kolom

Ket $|\psi\rangle$ direpresentasikan sebagai **vektor kolom**, sedangkan bra $\langle\psi|$ adalah **vektor baris** (konjugat transpos dari ket):

$$
|\psi\rangle\doteq\begin{bmatrix}c_H\\c_V\end{bmatrix}_{HV},
\qquad
\langle\psi|\doteq\begin{bmatrix}c_H^* & c_V^*\end{bmatrix}_{HV}.
$$

Inner product $\langle\psi_1|\psi_2\rangle$ dalam representasi ini adalah perkalian **vektor baris** $\times$ **vektor kolom**.

Sebagai contoh, proyeksi $\langle H|{+}45\rangle$:

$$
\langle H|{+}45\rangle
=
\begin{bmatrix}1 & 0\end{bmatrix}_{HV}
\frac{1}{\sqrt{2}}\begin{bmatrix}1\\1\end{bmatrix}_{HV}
=
\frac{1}{\sqrt{2}}.
$$

```python
print("Proyeksi <H|+45>")
result = inner_product(ket_H, ket_plus45)
print(f"<H|+45> = {clean_array(result)}")
print(f"Nilai analitik: 1/sqrt(2) = {1/np.sqrt(2):.6f}")
print(f"Apakah sama? {np.isclose(result, 1/np.sqrt(2))}")

print()
print("Tabel semua inner product antara keadaan basis")
states = {"H": ket_H, "V": ket_V, "+45": ket_plus45, "-45": ket_minus45, "L": ket_L, "R": ket_R}

names = list(states.keys())
print(f"{'':>6}", end="")
for name in names:
    print(f"{name:>14}", end="")
print()

for bra_name in names:
    print(f"{bra_name:>6}", end="")
    for ket_name in names:
        val = inner_product(states[bra_name], states[ket_name])
        val = complex(round(val.real, 4), round(val.imag, 4))
        print(f"{str(val):>14}", end="")
    print()
```

# Probabilitas dan Amplitudo Probabilitas

Koefisien $c_H$ dan $c_V$ disebut amplitudo probabilitas. Kuadrat magnitudonya memberikan probabilitas foton terukur berada pada keadaan tersebut:

$$
P\bigl(\langle H\big||\psi\rangle\bigr) = |c_H|^2 = |\langle H|\psi\rangle|^2,
$$

$$
P\bigl(\langle V\big||\psi\rangle\bigr) = |c_V|^2 = |\langle V|\psi\rangle|^2.
$$

Contoh: keadaan $|{+}45\rangle$ terurai menjadi $|H\rangle$ dan $|V\rangle$ dengan probabilitas yang sama,

$$
P\bigl(\langle H\big||{+}45\rangle\bigr) = \left|\frac{1}{\sqrt{2}}\right|^2 = \frac{1}{2},
\qquad
P\bigl(\langle V\big||{+}45\rangle\bigr) = \left|\frac{1}{\sqrt{2}}\right|^2 = \frac{1}{2}.
$$

Ini konsisten dengan hasil eksperimen: dari $N$ foton pada keadaan $|{+}45\rangle$, rata-rata $N/2$ terukur sebagai $|H\rangle$ dan $N/2$ sebagai $|V\rangle$.

```python
print("Probabilitas pengukuran dari keadaan |+45>")
P_H_from_plus45  = probability(ket_H, ket_plus45)
P_V_from_plus45  = probability(ket_V, ket_plus45)
print(f"P(<H| |+45>) = |<H|+45>|^2 = {P_H_from_plus45:.4f}")
print(f"P(<V| |+45>) = |<V|+45>|^2 = {P_V_from_plus45:.4f}")
print(f"Total probabilitas = {P_H_from_plus45 + P_V_from_plus45:.4f}")

print()
print("Probabilitas pengukuran dari keadaan |-45>")
P_H_from_minus45 = probability(ket_H, ket_minus45)
P_V_from_minus45 = probability(ket_V, ket_minus45)
print(f"P(<H| |-45>) = |<H|-45>|^2 = {P_H_from_minus45:.4f}")
print(f"P(<V| |-45>) = |<V|-45>|^2 = {P_V_from_minus45:.4f}")
print(f"Total probabilitas = {P_H_from_minus45 + P_V_from_minus45:.4f}")

print()
print("Probabilitas pengukuran dari keadaan |V> dalam basis +-45")
P_p45_from_V = probability(ket_plus45, ket_V)
P_m45_from_V = probability(ket_minus45, ket_V)
print(f"P(<+45| |V>) = {P_p45_from_V:.4f}")
print(f"P(<-45| |V>) = {P_m45_from_V:.4f}")
```

# Amplitudo Probabilitas Kompleks: Keadaan Melingkar $|L\rangle$ dan $|R\rangle$

Dengan analisis fase dari kondisi ortogonalitas dan hasil eksperimen, diperoleh

$$
|L\rangle = \frac{1}{\sqrt{2}}\bigl(|H\rangle + i|V\rangle\bigr),
\qquad
|R\rangle = \frac{1}{\sqrt{2}}\bigl(|H\rangle - i|V\rangle\bigr).
$$

Koefisien $i$ di sini adalah bilangan **imajiner**, membuktikan bahwa amplitudo probabilitas memang bilangan kompleks secara umum — bukan hanya bilangan riil.

Probabilitas pengukuran dari $|L\rangle$ dalam basis $HV$:

$$
P\bigl(\langle H\big||L\rangle\bigr) = \left|\frac{1}{\sqrt{2}}\right|^2 = \frac{1}{2},
\qquad
P\bigl(\langle V\big||L\rangle\bigr) = \left|\frac{i}{\sqrt{2}}\right|^2 = \frac{1}{2}.
$$

```python
print("Verifikasi |L> dan |R>")
ket_L_manual = (1/np.sqrt(2)) * (ket_H + 1j * ket_V)
ket_R_manual = (1/np.sqrt(2)) * (ket_H - 1j * ket_V)

print_ket("L (manual)", ket_L_manual)
print_ket("R (manual)", ket_R_manual)

print("Apakah |L> manual sama dengan definisi awal?")
print(np.allclose(ket_L_manual, ket_L))
print("Apakah |R> manual sama dengan definisi awal?")
print(np.allclose(ket_R_manual, ket_R))

print()
print("Ortogonalitas |L> dan |R>")
print(f"<L|R> = {clean_array(inner_product(ket_L, ket_R))}")
print(f"<R|L> = {clean_array(inner_product(ket_R, ket_L))}")

print()
print("Probabilitas pengukuran dari |L>")
print(f"P(<H| |L>) = {probability(ket_H, ket_L):.4f}")
print(f"P(<V| |L>) = {probability(ket_V, ket_L):.4f}")
print(f"P(<+45| |L>) = {probability(ket_plus45, ket_L):.4f}")
print(f"P(<-45| |L>) = {probability(ket_minus45, ket_L):.4f}")

print()
print("Probabilitas pengukuran dari |R>")
print(f"P(<H| |R>) = {probability(ket_H, ket_R):.4f}")
print(f"P(<V| |R>) = {probability(ket_V, ket_R):.4f}")
print(f"P(<+45| |R>) = {probability(ket_plus45, ket_R):.4f}")
print(f"P(<-45| |R>) = {probability(ket_minus45, ket_R):.4f}")
```

# Latihan Soal 1: Menghitung $\langle{-}45|L\rangle$

Hitunglah $\langle{-}45|L\rangle$ menggunakan representasi bracket maupun vektor baris-kolom.

**Metode Bracket:** Bra dari $|{-}45\rangle = \frac{1}{\sqrt{2}}(|H\rangle - |V\rangle)$ adalah

$$
\langle{-}45| = \frac{1}{\sqrt{2}}(\langle H| - \langle V|).
$$

Sehingga

$$
\langle{-}45|L\rangle
= \begin{bmatrix}\frac{1}{\sqrt{2}}(\langle H| - \langle V|)\end{bmatrix}\begin{bmatrix}\frac{1}{\sqrt{2}}(|H\rangle + i|V\rangle)\end{bmatrix}
$$

$$
\langle{-}45|L\rangle
= \frac{1}{2}(\langle H|H\rangle + i\langle H|V\rangle - \langle V|H\rangle - i\langle V|V\rangle)
= \frac{1}{2}(1 + 0 - 0 - i) = \frac{1-i}{2}.
$$

**Metode Vektor Baris-Kolom:**

$$
\langle{-}45|L\rangle
=
\begin{pmatrix}\frac{1}{\sqrt{2}}\begin{bmatrix}1 & -1\end{bmatrix}\end{pmatrix}
\begin{pmatrix}\frac{1}{\sqrt{2}}\begin{bmatrix}1\\i\end{bmatrix}\end{pmatrix}
= \frac{1}{2}(1 - i) = \frac{1-i}{2}.
$$

```python
print("Soal 1: Menghitung <-45|L>")
result = inner_product(ket_minus45, ket_L)
print(f"<-45|L> = {clean_array(result)}")
print(f"Nilai analitik: (1-i)/2 = {(1-1j)/2}")
print(f"Apakah sama? {np.isclose(result, (1-1j)/2)}")

print()
print("Probabilitas dari inner product ini")
print(f"|<-45|L>|^2 = {abs(result)**2:.4f}")
print("(Hasilnya Konsisten: P(<-45||L>) = 1/2)")
```

# Latihan Soal 2: Keadaan Polarisasi Eliptis $|e_1\rangle$

Keadaan polarisasi eliptis secara umum didefinisikan sebagai

$$
|e_1\rangle = \cos\theta\,|H\rangle + e^{i\phi}\sin\theta\,|V\rangle.
$$

**(i) Keadaan ortogonal $|e_2\rangle$:**

Dengan syarat $\langle e_1|e_2\rangle = 0$, diperoleh

$$
|e_2\rangle = \sin\theta\,|H\rangle - e^{i\phi}\cos\theta\,|V\rangle.
$$

**(ii) Menghitung $\langle{+}45|e_1\rangle$:**

$$
\langle{+}45|e_1\rangle
= \frac{1}{\sqrt{2}}(\langle H| + \langle V|)(\cos\theta\,|H\rangle + e^{i\phi}\sin\theta\,|V\rangle)
$$

$$
=\frac{1}{\sqrt{2}}(\cos\theta\, \langle H|H\rangle + e^{i\phi}\sin\theta\, \langle V|V\rangle)
$$
$$
= \frac{1}{\sqrt{2}}(\cos\theta + e^{i\phi}\sin\theta).
$$

```python
def ket_e1(theta_deg, phi_deg):
    return ket_elliptical(theta_deg, phi_deg)

def ket_e2(theta_deg, phi_deg):
    """Keadaan ortogonal terhadap e1: sin(theta)|H> - e^{i phi} cos(theta)|V>."""
    theta = np.deg2rad(theta_deg)
    phi   = np.deg2rad(phi_deg)
    return np.array([[np.sin(theta)], [-np.exp(1j*phi)*np.cos(theta)]], dtype=complex)

theta_test = 30
phi_test   = 60

e1 = ket_e1(theta_test, phi_test)
e2 = ket_e2(theta_test, phi_test)

print(f"Uji dengan theta = {theta_test} deg, phi = {phi_test} deg")
print()
print_ket("e1", e1)
print_ket("e2", e2)

print("(i) Verifikasi ortogonalitas <e1|e2> = 0")
ip_e1e2 = inner_product(e1, e2)
print(f"<e1|e2> = {clean_array(ip_e1e2)}")
print(f"Apakah ortogonal? {np.isclose(ip_e1e2, 0)}")

print()
print("Verifikasi normalisasi")
print(f"<e1|e1> = {inner_product(e1, e1).real:.6f}")
print(f"<e2|e2> = {inner_product(e2, e2).real:.6f}")

print()
print("(ii) Menghitung <+45|e1>")
result_bracket = inner_product(ket_plus45, e1)
theta_r = np.deg2rad(theta_test)
phi_r   = np.deg2rad(phi_test)
analytic = (1/np.sqrt(2)) * (np.cos(theta_r) + np.exp(1j*phi_r)*np.sin(theta_r))
print(f"<+45|e1> (numerik)   = {clean_array(result_bracket)}")
print(f"<+45|e1> (analitik)  = {clean_array(analytic)}")
print(f"Apakah sama? {np.isclose(result_bracket, analytic)}")
```

# Latihan Soal 3: Representasi $|{\pm}45\rangle$ dalam Basis $LR$

Koefisien dalam basis $\{|L\rangle, |R\rangle\}$ diperoleh dengan proyeksi:

$$
c_L = \langle L|\psi\rangle, \qquad c_R = \langle R|\psi\rangle.
$$

Untuk $|{+}45\rangle$:

$$
c_L = \langle L|{+}45\rangle = \frac{1-i}{2}, \qquad c_R = \langle R|{+}45\rangle = \frac{1+i}{2},
$$

sehingga

$$
|{+}45\rangle_C \doteq \frac{1}{2}\begin{bmatrix}1-i\\1+i\end{bmatrix}_C.
$$

Untuk $|{-}45\rangle$:

$$
c_L = \langle L|{-}45\rangle = \frac{1+i}{2}, \qquad c_R = \langle R|{-}45\rangle = \frac{1-i}{2},
$$

sehingga

$$
|{-}45\rangle_C \doteq \frac{1}{2}\begin{bmatrix}1+i\\1-i\end{bmatrix}_C.
$$

```python
print("oal 3: Representasi |+45> dan |-45> dalam basis LR")

# Koefisien untuk |+45>
c_L_plus  = inner_product(ket_L, ket_plus45)
c_R_plus  = inner_product(ket_R, ket_plus45)

# Koefisien untuk |-45>
c_L_minus = inner_product(ket_L, ket_minus45)
c_R_minus = inner_product(ket_R, ket_minus45)

print("Vektor kolom |+45> dalam basis LR:")
print(f"  c_L = <L|+45> = {clean_array(c_L_plus)}")
print(f"  c_R = <R|+45> = {clean_array(c_R_plus)}")
print(f"  |+45>_C = [{c_L_plus:.4f}, {c_R_plus:.4f}]^T")

print()
print("Nilai analitik: (1-i)/2 dan (1+i)/2")
print(f"  (1-i)/2 = {(1-1j)/2},  (1+i)/2 = {(1+1j)/2}")
print(f"  Apakah c_L sama? {np.isclose(c_L_plus, (1-1j)/2)}")
print(f"  Apakah c_R sama? {np.isclose(c_R_plus, (1+1j)/2)}")

print()
print("Vektor kolom |-45> dalam basis LR:")
print(f"  c_L = <L|-45> = {clean_array(c_L_minus)}")
print(f"  c_R = <R|-45> = {clean_array(c_R_minus)}")
print(f"  |-45>_C = [{c_L_minus:.4f}, {c_R_minus:.4f}]^T")

print()
print("Verifikasi rekonstruksi ")
recon_plus  = c_L_plus  * ket_L + c_R_plus  * ket_R
recon_minus = c_L_minus * ket_L + c_R_minus * ket_R
print("Apakah rekonstruksi |+45> dari basis LR benar?")
print(np.allclose(recon_plus, ket_plus45))
print("Apakah rekonstruksi |-45> dari basis LR benar?")
print(np.allclose(recon_minus, ket_minus45))
```

# Latihan Soal 4 & 5: Keadaan $|\theta\rangle$ dan Probabilitasnya

**Soal 4:** Keadaan foton terpolarisasi linier pada sudut $\theta$ dari horizontal, dengan analogi dari fisika klasik:

$$
|\theta\rangle = \cos\theta\,|H\rangle + \sin\theta\,|V\rangle.
$$

**Soal 5(i):** Probabilitas terukur sebagai polarisasi vertikal:

$$
P\bigl(\langle V\big||\theta\rangle\bigr) = |\langle V|\theta\rangle|^2 = \sin^2\theta.
$$

**Soal 5(ii):** Probabilitas terukur sebagai polarisasi melingkar kanan:

$$
P\bigl(\langle R\big||\theta\rangle\bigr)
= \left|\frac{1}{\sqrt{2}}(\cos\theta + i\sin\theta)\right|^2
= \frac{1}{2}|e^{i\theta}|^2
= \frac{1}{2}.
$$

Menariknya, probabilitas ini **selalu** $\frac{1}{2}$ untuk sembarang sudut $\theta$. Foton pada keadaan polarisasi linier berarah apapun selalu memiliki peluang 50% terukur sebagai polarisasi melingkar kanan.

```python
print("Soal 4: Keadaan |theta> dalam basis HV")
theta_list = [0, 30, 45, 60, 90]

for theta in theta_list:
    ket = ket_theta(theta)
    print_ket(f"theta={theta} deg", ket)

print()
print("Soal 5(i): P(<V| |theta>) = sin^2(theta)")
print(f"{'theta (deg)':>12} | {'P(<V| |theta>)':>16} | {'sin^2(theta)':>14} | {'sama?':>6}")
print("-" * 56)
for theta in theta_list:
    ket = ket_theta(theta)
    p_num = probability(ket_V, ket)
    p_ana = np.sin(np.deg2rad(theta))**2
    print(f"{theta:>12} | {p_num:>16.6f} | {p_ana:>14.6f} | {np.isclose(p_num, p_ana)!s:>6}")

print()
print("Soal 5(ii): P(<R| |theta>) = 1/2 untuk semua theta")
theta_extended = [0, 15, 30, 45, 60, 75, 90, 120, 150, 180]
print(f"{'theta (deg)':>12} | {'P(<R| |theta>)':>16}")
print("-" * 32)
for theta in theta_extended:
    ket = ket_theta(theta)
    p = probability(ket_R, ket)
    print(f"{theta:>12} | {p:>16.6f}")
```

# Rangkuman Keadaan Kuantum Polarisasi

## Ket dan Bra

Polarisasi horizontal dan vertikal:

$$
|H\rangle = \begin{bmatrix}1\\0\end{bmatrix}, \qquad |V\rangle = \begin{bmatrix}0\\1\end{bmatrix}.
$$

Polarisasi linier $\pm 45^\circ$:

$$
|{+}45\rangle = \frac{1}{\sqrt{2}}(|H\rangle + |V\rangle), \qquad |{-}45\rangle = \frac{1}{\sqrt{2}}(|H\rangle - |V\rangle).
$$

Polarisasi melingkar:

$$
|L\rangle = \frac{1}{\sqrt{2}}(|H\rangle + i|V\rangle), \qquad |R\rangle = \frac{1}{\sqrt{2}}(|H\rangle - i|V\rangle).
$$

## Inner Product dan Probabilitas

Inner product antara bra $\langle\psi_1|$ dan ket $|\psi_2\rangle$:

$$
\langle\psi_1|\psi_2\rangle \in \mathbb{C}, \qquad \langle\psi_1|\psi_2\rangle^* = \langle\psi_2|\psi_1\rangle.
$$

Probabilitas mengukur $|\psi\rangle$ pada keadaan $|\phi\rangle$:

$$
P = |\langle\phi|\psi\rangle|^2.
$$

Kondisi normalisasi: $\langle\psi|\psi\rangle = 1$.
